```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
songs = pd.read_csv('filtered/songs_2021-2023.csv')
len(songs)
```

```python
song_index = 0

test_song = songs.iloc[song_index]
test_song['lyrics'] = test_song['lyrics'].encode('utf-8').decode('unicode_escape')
print(test_song['lyrics'])
```

```python
print('Valence: ' + str(test_song['valence_norm']))
sentiment = analyzer.polarity_scores(test_song['lyrics'])['compound']
print('Sentiment: ' + str(sentiment))
```
